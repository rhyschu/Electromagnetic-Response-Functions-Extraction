```python
import subprocess
import pandas as pd
import numpy as np
import os
import time
```

```python
qv_values = [0.1, 0.148, 0.167, 0.205, 0.24, 0.3, 0.38, 0.475, 0.57, 0.649, 0.756, 0.991, 1.619, 1.921, 2.213, 2.5, 2.783, 3.5]
q2_values = [0.001, 0.01, 0.02, 0.026, 0.04, 0.056, 0.093, 0.12, 0.16, 0.265, 0.38, 0.5, 0.8, 1.25, 1.75, 2.25, 2.75, 3.25, 3.75]
response_columns = ['qv','q2','ex','nu','RTTOT','RLTOT','RTQE','RLQE','RTIE','RLIE','RTE','RLE','RTNS','RLNS']
```

```python
os.makedirs("Qvedges", exist_ok=True)
os.makedirs("Q2edges", exist_ok=True)
```

```python
for qv in qv_values:
    df = pd.DataFrame({'nu': np.r_[np.arange(0, 1, 0.0001), np.arange(1, 3.6, 0.001)]})
    df['qv'] = qv
    df[['qv','nu']].to_csv('responseqv_input.txt', index=True, header=False, sep=' ')
    with open('responseqv_output.txt', 'w') as output_file:
        subprocess.run(['response/responseqv.exe', 'responseqv_input.txt'], stdout=output_file) 
    time.sleep(0.5)
    df = pd.read_csv('responseqv_output.txt', sep='\s+', header=None, names=response_columns)
    df = df.dropna()
    df.to_csv(f"Qvedges/Qvedge_{qv}.csv", index=False)
    
for q2 in q2_values:
    df = pd.DataFrame({'nu': np.r_[np.arange(0, 1, 0.0001), np.arange(1, 3.6, 0.001)]})
    df['q2'] = q2
    df[['q2','nu']].to_csv('responseq2_input.txt', index=True, header=False, sep=' ')
    with open('responseq2_output.txt', 'w') as output_file:
        subprocess.run(['response/responseq2.exe', 'responseq2_input.txt'], stdout=output_file) 
    time.sleep(0.5)
    df = pd.read_csv('responseq2_output.txt', sep='\s+', header=None, names=response_columns)
    df = df.dropna()
    df.to_csv(f"Q2edges/Q2edge_{q2}.csv", index=False)
```
